```python
def clean_text(text):
    if isinstance(text, str):   # process only strings
        text = text.lower()
        text = re.sub(r"http\S+|www\S+|https\S+", '', text)
        text = re.sub(r'@\w+|\#','', text)
        text = re.sub(r'[^a-zA-Z\s]', '', text)
        text = re.sub(r'\s+', ' ', text).strip()
        return text
    return ""   # if not string, return empty text

df['cleaned_statement'] = df['statement'].apply(clean_text)
```

```python
from nltk.corpus import stopwords

# Make sure stopwords are downloaded
import nltk
nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

# Remove stopwords from the 'tokens' column
df['filtered_tokens'] = df['tokens'].apply(lambda x: [word for word in x if word.lower() not in stop_words])
```

```python
# Display first 5 rows normally
print(df[['statement', 'tokens', 'filtered_tokens']].head())
```

```python
import textwrap

# Function to wrap long lists (tokens) into multiple lines
def wrap_list(lst, width=8):  # width = number of tokens per line
    lines = []
    for i in range(0, len(lst), width):
        lines.append(str(lst[i:i+width]))
    return "\n".join(lines)
```

```python
# Display first 5 rows with wrapped text and tokens
for i in range(5):
    print("Original statement:\n", "\n".join(textwrap.wrap(df['statement'][i], 80)))
    print("Tokens:\n", wrap_list(df['tokens'][i], 8))
    print("Filtered Tokens (after removing stopwords):\n", wrap_list(df['filtered_tokens'][i], 8))
    print("-"*100)
```

```python
import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Download required NLTK data
nltk.download('wordnet')
nltk.download('omw-1.4')

# Initialize stemmer and lemmatizer
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

# Apply stemming and lemmatization
df['stemmed'] = df['filtered_tokens'].apply(lambda x: [stemmer.stem(word) for word in x])
df['lemmatized'] = df['filtered_tokens'].apply(lambda x: [lemmatizer.lemmatize(word) for word in x])
```

```python
import re

# Function to normalize repeated characters
def normalize_text(tokens):
    return [re.sub(r'(.)\1{2,}', r'\1', word) for word in tokens]

# Apply normalization to lemmatized tokens
df['normalized'] = df['lemmatized'].apply(lambda x: normalize_text(x))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Convert the normalized tokens back to string for TF-IDF
df['cleaned_text'] = df['normalized'].apply(lambda x: ' '.join(x))

# Initialize TF-IDF vectorizer
tfidf = TfidfVectorizer(max_features=5000)

# Fit and transform the cleaned text
X = tfidf.fit_transform(df['cleaned_text']).toarray()

print("TF-IDF feature matrix shape:", X.shape)

# Optional: see first 10 feature names
print("First 10 TF-IDF features:", tfidf.get_feature_names_out()[:10])
```
